```python
attendance_work = attendance[
    [
        "No",
        "Date",
        "Employee No",
        "Employee Name",
        "Division",
        "Attendance Status",
        "Time In",
        "Time Out",
        "Late In",
        "Remarks",
    ]
].copy()
```

```python
#count day

pattern = "Present|Business Trip|Present Off|Absent|Leave"

attendance_work["Count Day"] = (
    attendance_work["Attendance Status"]
    .fillna("")
    .str.contains(pattern, case=False, regex=True)
    .astype(int)
)
```

```python
#clock in

attendance_work["Clock In"] = (
    attendance_work["Time In"].notna()
    |
    attendance_work["Attendance Status"]
        .fillna("")
        .str.contains("Leave", case=False)
).astype(int)
```

```python
#clock out

attendance_work["Clock Out"] = (
    attendance_work["Time Out"].notna()
    |
    attendance_work["Attendance Status"]
        .fillna("")
        .str.contains("Leave", case=False)
).astype(int)
```

```python
#late

attendance_work["Late"] = (
    attendance_work["Late In"] > 0
).astype(int)
```

```python
summary = (
    attendance_work
    .groupby(["Employee No", "Employee Name"])
    .agg(
        Division=("Division", "first"),
        Count_Day=("Count Day", "sum"),
        Clock_In=("Clock In", "sum"),
        Clock_Out=("Clock Out", "sum"),
        Late=("Late", "sum")
    )
    .reset_index()
)
```

```python
summary["C.In"] = np.where(
    summary["Count_Day"] == 0,
    0,
    summary["Clock_In"] / summary["Count_Day"]
)

summary["C.Out"] = np.where(
    summary["Count_Day"] == 0,
    0,
    summary["Clock_Out"] / summary["Count_Day"]
)

summary["Late"] = np.where(
    summary["Count_Day"] == 0,
    0,
    summary["Late"] / summary["Count_Day"]
)
```

```python
summary["C.In"] = summary["C.In"].map("{:.0%}".format)
summary["C.Out"] = summary["C.Out"].map("{:.0%}".format)
summary["Late"] = summary["Late"].map("{:.0%}".format)
```

```python
import pandas as pd

# Membaca file
attendance = pd.read_excel("../data/Attendance_Juli.xlsx")

# Membersihkan spasi pada nama divisi
attendance["Division"] = attendance["Division"].astype(str).str.strip()

# Mengambil semua divisi yang mengandung kata "Asset"
attendance = attendance[
    attendance["Division"].str.contains("Asset", case=False, na=False)
].copy()

print("Attendance:", attendance.shape)

# Memilih kolom yang dibutuhkan
attendance_work = attendance[
    [
        "No",
        "Date",
        "Employee No",
        "Employee Name",
        "Division",
        "Attendance Status",
        "Time In",
        "Time Out",
        "Late In",
        "Remarks"
    ]
].copy()

print("Attendance Work:", attendance_work.shape)

display(attendance_work.head(300))
```
